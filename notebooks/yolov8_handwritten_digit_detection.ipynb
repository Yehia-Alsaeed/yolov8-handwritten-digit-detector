# YOLOv8 Handwritten Digit Detection

This notebook trains and evaluates a YOLOv8 object detection model for localizing and classifying handwritten digit instances from the HashiDigits dataset.

## Project Goal

Detect handwritten digit instances in images using transfer learning with YOLOv8n, then test the trained model on dataset samples and real handwritten inputs.

## Libraries

```python
import os
import cv2
import random
import matplotlib.pyplot as plt
from ultralytics import YOLO
from IPython.display import display
from PIL import Image
import pandas as pd
```

## Loading Dataset

```python
path = "data.yaml"
model = YOLO("yolov8n.pt")
```

```python
os.environ['KMP_DUPLICATE_LIB_OK']='TRUE'
```

```python
# Training The Model
results = model.train(data=path, epochs=10, imgsz=416)
```

```python
model = YOLO(r"runs\detect\train2\weights\last.pt")
results = model.train(data=path, epochs=10, imgsz=416)
```

```python
model = YOLO(r"runs\detect\train3\weights\last.pt")
results = model.train(data=path, epochs=20, imgsz=416)
```

```python
# Training again with bigger image size
model = YOLO(r"runs\detect\train4\weights\last.pt")
results = model.train(data=path, epochs=20, imgsz=640, augment=True)
```

```python
model = YOLO(r"runs\detect\train5\weights\last.pt")  
results = model.train(resume=True)
```

```python
# Load the trained checkpoint included with this repository
model = YOLO("models/best.pt") 
```

```python
results = model([r"examples/small-handwriting-sample.jpeg"],iou=0)  
```

# Testing The Model

```python
results = model([r"examples/hashidigits-sample.jpg"], iou=0)

# Loop through results and display predictions
for i, result in enumerate(results):
    result.save(filename=f"pred_{i}.jpg")  # Save prediction image
    display(Image.open(f"pred_{i}.jpg"))   # Show it in notebook
```

```python
results = model([r"examples/multi-digit-sample.jpg"], iou=0)

for i, result in enumerate(results):
    result.save(filename=f"pred_{i}.jpg")  
    display(Image.open(f"pred_{i}.jpg"))  
```

```python
results = model([r"examples/small-handwriting-sample.jpeg"], iou=0)

for i, result in enumerate(results):
    result.save(filename=f"pred_{i}.jpg")  # Save prediction image
    display(Image.open(f"pred_{i}.jpg"))   # Show it in notebook
```

# Testing The Model on Real Life Handwriting

```python
results = model([r"examples/real-handwriting-7.jpeg"], iou=0)

for i, result in enumerate(results):
    result.save(filename=f"pred_{i}.jpg")  
    display(Image.open(f"pred_{i}.jpg"))  
```

# Visualization

```python
%matplotlib inline

results_path = r"runs\detect\train5\results.csv"
results = pd.read_csv(results_path)

plt.figure(figsize=(8, 5))
plt.plot(results['epoch'], results['metrics/mAP50(B)'], 'b-', linewidth=2)
plt.title('Model Accuracy Over Training')
plt.xlabel('Epoch')
plt.ylabel('mAP50(B) (Accuracy)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
